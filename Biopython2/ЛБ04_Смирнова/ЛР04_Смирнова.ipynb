## Создайте функцию для записи последовательностей из строк в файл вормата .fasta. C помощью этой функции запишите последовательность из переменной seq1 в fasta-файл. Идентификатор и описание для последовательности придумайте самостоятельно. 

```python
seq1= 'ATAGCGACTACCCCTGGCAACCGCGAAGCTCTGCGGTCCCGCGGTCGGGCTACGGGTTTGAGCAAAGCTCCTCTCTTCCCTTCACTTCCCTCCGGACTGGTTTCTTCTTCCTTCCCCCTTCCCCCAACTTCCCTCCACCCCTTCCA'
```

```python
def write_fasta(filename, id, seq, l = 100):
    with open(filename,"w") as f:
        f.write(f">{id}\n")
        for i in range(0,len(seq),l):
            f.write(seq[i:i+l]+"\n")
    print(f"Последовательность записана в файл {filename}")

write_fasta(
    filename="seq1_out.fasta",
    id="seq1",
    seq=seq1
)
```

## Выведите содержимое файла "homologene.fa". Сначала словестно охарактеризйте его содержимое. После чего проанализируйте все последовательности файла с помощью Biopython: рассчитайте длину, gc-состав, молекулярную массу, нуклеотидный состав и найдите позиции старт-кодонов. 

```python
Файл содержит 7 последовательностей от различных организмов: Homo sapines, Cebus capucinus,
Felis catus,  Trionyx triunguis, Halicephalobus mephisto, Drosophila melanogaster, Babyrousa celebensis
Все последовательности имеют схожую длину.
```

```python
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction, molecular_weight

rec = list(SeqIO.parse("homologene.fa", "fasta"))
with open("homologene.fa", "r") as f:
    for line in f:
        print(line, end="")
rec
```

```python
for r in rec:
    seq = r.seq
    seq = str(seq)
    l = len(seq)
    gc = gc_fraction(seq) * 100
    mol_weight = molecular_weight(seq, seq_type="DNA")
    A = seq.count('A')
    T = seq.count('T')
    G = seq.count('G')
    C = seq.count('C')

    start_codons = []
    for i in range(len(seq) - 2):
        codon = seq[i:i+3]
        if codon == 'ATG':
            start_codons.append(i + 1)

    print(f"> {r.description}")
    print(f"длина = {l}, gc-состав = {gc:.2f}, молекулярная масса = {mol_weight:.2f}")
    print(f"A = {A}, T = {T}, G = {G}, C = {C}")
    print(f"позиция старт кодона = {start_codons}")
    print(" "*60)
```

# Phylogenetic Tree

```python
from Bio import Phylo
from Bio.Phylo.TreeConstruction import DistanceCalculator
from Bio import AlignIO

import pylab
```

## 1.1 Загрузите файл с ридами гена и сохраните в новый формат. Используйте with open, чтобы открыть файл. AlignIO.parse - сохранить данные в переменную и AlignIO.write - записать данные в новый файл с расширением .phy.

```python
with open("homologene.fa", "r") as f:
    seq = AlignIO.read(f,"fasta")

    with open("out.phy","w") as ff:
        AlignIO.write(seq, ff, "phylip")
print("Файл конвертирован")
```

## 1.2 Посчитайте дистанцию между генами с помощью объекта DistanceCalculator. И его метода calculator.get_distance

```python
from Bio.Phylo.TreeConstruction import DistanceCalculator, DistanceTreeConstructor
```

```python
calculator = DistanceCalculator('identity')
dm = calculator.get_distance(seq)
print(dm)
```

## 1.3 Постройте дерево с помощью DistanceTreeConstructor / constructor.build_tree и сделайте по нему вывод

```python
constructor = DistanceTreeConstructor(calculator, 'upgma')
tree = constructor.build_tree(seq)

Phylo.draw_ascii(tree)
Phylo.write(tree, "tree.nwk", "newick")
print("Дерево в tree.nwk")
```
