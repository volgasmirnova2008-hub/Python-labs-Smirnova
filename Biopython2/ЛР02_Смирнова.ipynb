## Создайте класс SmartHouse (умный дом), отразив при этом основные принципы ООП. Поясните логику своего кода. 

```python
class Dev:
    def __init__(self, name):
        self.name = name
        self.on = False

    def turn_on(self):
        self.on = True
        print(f" {self.name} включен")

    def turn_off(self):
        self.on = False
        print(f" {self.name} выключен")

class Light(Dev):
    def __init__(self, name):
        super().__init__(name)       
        self.brightness = 0          

    def turn_on(self):
        super().turn_on()            
        self.brightness = 100
        print(f" Яркость: {self.brightness}%")

    def turn_off(self):
        super().turn_off()            
        self.brightness = 0
        print(f" Яркость: {self.brightness}%")


class Thermo(Dev):
    def __init__(self, name):
        super().__init__(name)
        self.temperature = 25

    def turn_on(self):
        super().turn_on()
        print(f" Температура: {self.temperature}°C")

    def turn_off(self):
        super().turn_off()
        print(f" Температура: {self.temperature}°C")

class SmartHouse:
    def __init__(self):
        self._devices = []      

    def add_device(self, device):
        self._devices.append(device)

    def turn_on_all(self):         
        print("Включаем все устройства:")
        for device in self._devices:
            device.turn_on()      

    def turn_off_all(self):         
            print("Выключаем все устройства:")
            for device in self._devices:
                device.turn_off()  

    def show_status(self):
        print("Статус устройств:")
        for device in self._devices:
            state = "включена" if device.on else "выключена"
            print(f"  {device.name}: {state}")


house = SmartHouse()

ligh= Light("Люстра")
house.add_device(ligh)
house.add_device(Light("Лампа"))
house.add_device(Thermo("Батарея"))



house.turn_on_all()
house.show_status()
house.turn_off_all()

house.show_status()

ligh.turn_on()
house.show_status()
```

```python
def cldna(data):
    data = data.strip().upper()
    data_n = []
    for el in data:
        if el in "ATGC":
            data_n.append(el)
    n = 3
    finalf = []
    for i in range(0, len(data_n), n):
        final = data_n[i: i+n]
        finalf.append("".join(final))
    return " ".join(finalf)
```

## Для полученной последовательности вычислите её молекулярную массу, температуру плавления, GC-состав, а также длину цепи

```python
def mol_weight(dna_str):
    
    pattern_dict = {
        "A": 313.21,
        "G": 329.21,
        "T": 304.2,
        "C": 289.18
    }
    
    const = 61.96 
    
    dna_str = dna_str.upper()
    weight = 0
    
    for nuc in dna_str:
        if nuc in pattern_dict:
            weight += pattern_dict[nuc.upper()]
    return round(weight - const, 2)

mol_weight(dna)
```

```python
def temperature(dna_str):
    
    g_count = dna_str.upper().count("G")
    c_count = dna_str.upper().count("C")
    t_count = dna_str.upper().count("T")
    a_count = dna_str.upper().count("A")
    
    if len(dna_str) <= 13:

        return (t_count+a_count)*2 + (c_count+g_count)*4
    
    else:
        
        return int(64.9 +41*(g_count+c_count-16.4)/(a_count+t_count+g_count+c_count))

temperature(dna)
```

```python
def cldna2(data):
    data = data.strip().upper()
    data_n = []
    for el in data:
        if el in "ATGC":
            data_n.append(el)
    return "".join(data_n)

dna2 = cldna2(raw_seq)
            
def concentration(dna_str):
    
    g_conc = dna_str.upper().count("G")
    c_conc = dna_str.upper().count("C")
    
    return round((g_conc+c_conc) * 100 / len(dna_str), 0)

concentration(dna2)
```

## Из полученной нуклеотидной последовательности из 2.1 сделайте матричную-РНК

```python
def transcriptor(dna_str):

    pattern_dict = {
        "A": "U",
        "G":"C",
        "T":"A",
        "C":"G"
    }
    
    rna_str = ""
    
    for nuc in dna_str:
        if nuc in pattern_dict:
            rna_str += pattern_dict[nuc.upper()]
        
    return rna_str
```

## Через Biopython реализуйте функцию, которая принимает строку с матричной РНК (из прошлого задания 2.3) и преобразует её в список аминокислот, соответствующих каждому триплету. При этом функция должна начинать считывание с первого встреченного старт-кодона и завершать работу при обнаружении стоп-кодона

```python
from Bio.Seq import Seq

def translate1(mrna_1):
    mrna = Seq(mrna_1.replace(" ", ""))
    start_pos = mrna.find("AUG")
    if start_pos == -1:
        return [] 
    protein = mrna[start_pos:].translate(to_stop= True)
    return list(protein)

translate1(mrna)
```
